## Question 1 — Broadcasting and Indexing

You are given:
```python
import numpy as np
arr = np.array([[2, 4, 6],
 [8, 10, 12],
 [14, 16, 18]])
 ```
1. Use slicing to extract the second and third columns.
2. Add a NumPy array [1, 2, 3] to each row using broadcasting.
3. Explain how broadcasting works in this context

````python
arr = np.array([[2, 4, 6],
 [8, 10, 12],
 [14, 16, 18]])
````

````python
# 1. Use slicing to extract the second and third columns.
second_col = arr[:, 1]
third_col = arr[:, 2]

second_col, third_col
````

````python
# 2. Add a NumPy array [1, 2, 3] to each row using broadcasting.
arr2 = np.array([1, 2, 3])
arr_total = arr + arr2

arr_total
````

3. Explain how broadcasting works in this context

- arr has shape (3, 3)
- arr2 has shape (3,)

> NumPy automatically expands the 1D array [1, 2, 3] to match the shape of the 2D array (3, 3) along the rows. Each row of arr is added elementwise to [1, 2, 3]. This is called broadcasting, which allows vectorized operations between arrays of different but compatible shapes.

## Question 2 — Random and Statistical Functions

Create a 4×4 array of random integers between 10 and 50 (inclusive).

Then:
1. Find the mean, standard deviation, and maximum of the array.
2. Replace all elements greater than the mean with 0.
3. Print the modified array.
(Hint: use np.random.randint, np.mean, and boolean indexing)

````python
# Create a 4×4 array of random integers between 10 and 50 (inclusive)
arr = np.random.randint(low=10, high=51, size=(4, 4))
arr
````

````python
# 1. Find the mean, standard deviation, and maximum of the array.

arr_mean = np.mean(arr)
print("Array mean :", arr_mean)

arr_std = np.std(arr)
print("Array standard deviation :", arr_std)

arr_max = np.max(arr)
print("Array max value :", arr_max)
````

````python
# 2. Replace all elements greater than the mean with 0.
arr_bool = arr > arr_mean
arr[arr_bool] = 0
````

````python
# 3. Print the modified array.
print(arr)
````

## Question 3 — Matrix Operations

Given the matrix:
```python
A = np.array([[4, 2],
              [3, 5]])
```

1. Compute the transpose and inverse of A.
2. Verify that the product of A and its inverse gives an identity matrix (within rounding error).
3. Explain why np.dot() is preferred over the * operator for matrix multiplication.

````python
A = np.array([[4, 2],
              [3, 5]])
````

````python
# 1. Compute the transpose and inverse of A.

A_transpose = A.T
A_transpose
````

````python
# 2. Verify that the product of A and its inverse gives an identity matrix (within rounding error).

A_inverse = np.linalg.inv(A)
output_matrix = np.dot(A, A_inverse)
output_matrix = np.round(output_matrix, 6)
output_matrix
````

3. Explain why np.dot() is preferred over the * operator for matrix multiplication.

- `A * B` performs elementwise multiplication of the corresponding elemenets.
- While `np.dot(A, B)` performs true matrix multiplication according to the mathemtical definition.

## Question 4 — linspace, reshape, and Slicing

1. Create an array of 15 evenly spaced values between 0 and 30 using np.linspace().
2. Reshape it into a 3×5 matrix.
3. Extract the last two rows and the first three columns from the reshaped array.
4. Compute the column-wise mean of this subarray.

````python
# 1. Create an array of 15 evenly spaced values between 0 and 30 using np.linspace()
arr = np.linspace(0, 30, num=15)
arr
````

````python
# 2. Reshape it into a 3×5 matrix.
arr_reshaped = arr.reshape(3, 5)
arr_reshaped
````

````python
# 3. Extract the last two rows and the first three columns from the reshaped array.
extracted_arr = arr_reshaped[-2:, :3]
extracted_arr
````

````python
# 4. Compute the column-wise mean of this subarray.

arr_mean_colwise = np.mean(extracted_arr, axis=0)
arr_mean_colwise
````
